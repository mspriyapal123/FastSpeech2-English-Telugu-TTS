```python
import os, csv, subprocess, soundfile as sf
import numpy as np
from scipy.signal import resample_poly
from math import gcd

base = "/content/drive/MyDrive/FastSpeech2_Internship/Telugu"
raw = os.path.join(base, "raw_data")
out = os.path.join(base, "processed")
audio_out = os.path.join(out, "audio")

os.makedirs(audio_out, exist_ok=True)

success = 0
failed = []

with open(os.path.join(raw, "metadata.csv"), encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

with open(os.path.join(out, "metadata.csv"), "w",
          newline="", encoding="utf-8") as f:

    writer = csv.writer(f)
    writer.writerow(["filename", "text", "phonemes"])

    for row in rows:
        try:
            filename = row["filename"]
            text = row["text"]

            audio, sr = sf.read(os.path.join(raw, filename))

            if audio.ndim > 1:
                audio = audio.mean(axis=1)

            if sr != 22050:
                g = gcd(sr, 22050)
                audio = resample_poly(
                    audio, 22050 // g, sr // g
                )

            new_name = os.path.splitext(filename)[0] + ".wav"
            sf.write(
                os.path.join(audio_out, new_name),
                audio, 22050
            )

            result = subprocess.run(
                ["espeak-ng", "-q", "-v", "te",
                 "-x", "--sep=_", text],
                capture_output=True,
                text=True,
                check=True
            )

            phonemes = result.stdout.strip()

            if not phonemes:
                raise ValueError("Empty phoneme output")

            writer.writerow([new_name, text, phonemes])
            success += 1

        except Exception as e:
            failed.append((row["filename"], str(e)))

print("Successfully processed:", success)
print("Failed:", len(failed))
print("Output folder:", out)

if failed:
    print("First errors:", failed[:5])
```

```python
import os
import csv
import numpy as np
import soundfile as sf
import librosa

english_path = "/content/drive/MyDrive/FastSpeech2_Internship/English/raw_data"
os.makedirs(english_path, exist_ok=True)

metadata = []
saved = 0
failed = 0

for sample in english_ds:
    if saved >= 100:
        break

    try:
        # Decode audio
        decoded = sample["audio_filepath"].get_all_samples()
        audio = decoded.data.cpu().numpy()
        sr = decoded.sample_rate

        # Convert to mono
        if audio.ndim == 2:
            audio = audio.mean(axis=0)

        # Resample to 22050 Hz
        if sr != 22050:
            audio = librosa.resample(
                audio.astype(np.float32),
                orig_sr=sr,
                target_sr=22050
            )

        text = str(sample["text"]).strip()
        if not text:
            continue

        filename = f"english_{saved:04d}.wav"
        sf.write(os.path.join(english_path, filename), audio, 22050)

        metadata.append([filename, text])
        saved += 1

        if saved % 20 == 0:
            print("Saved:", saved)

    except Exception as e:
        failed += 1
        if failed <= 3:
            print("Error:", str(e)[:200])

with open(
    os.path.join(english_path, "metadata.csv"),
    "w",
    newline="",
    encoding="utf-8"
) as f:
    writer = csv.writer(f)
    writer.writerow(["filename", "text"])
    writer.writerows(metadata)

print("\nSuccessfully saved:", saved)
print("Failed:", failed)
print("Folder:", english_path)
```

```python
import os
import csv
import re
import subprocess
import numpy as np
import soundfile as sf

base = "/content/drive/MyDrive/FastSpeech2_Internship/English"
raw_dir = os.path.join(base, "raw_data")
processed_dir = os.path.join(base, "processed")
audio_dir = os.path.join(processed_dir, "audio")

os.makedirs(audio_dir, exist_ok=True)

processed = []
failed = 0

with open(os.path.join(raw_dir, "metadata.csv"), encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

for row in rows:
    try:
        filename = row["filename"]
        text = re.sub(r"\s+", " ", row["text"]).strip()

        audio, sr = sf.read(os.path.join(raw_dir, filename))

        if audio.ndim == 2:
            audio = audio.mean(axis=1)

        # Audio already resampled to 22050 Hz
        if sr != 22050:
            raise ValueError(f"Unexpected sample rate: {sr}")

        sf.write(os.path.join(audio_dir, filename), audio, sr)

        # Generate English phonemes
        result = subprocess.run(
            ["espeak-ng", "-q", "-v", "en-us", "-x", "--sep=_", text],
            capture_output=True,
            text=True,
            check=True
        )

        phonemes = result.stdout.strip().replace("\n", " ")

        if not phonemes:
            raise ValueError("Empty phoneme sequence")

        processed.append([filename, text, phonemes])

    except Exception as e:
        failed += 1
        if failed <= 3:
            print("Error:", str(e))

with open(
    os.path.join(processed_dir, "metadata.csv"),
    "w",
    newline="",
    encoding="utf-8"
) as f:
    writer = csv.writer(f)
    writer.writerow(["filename", "text", "phonemes"])
    writer.writerows(processed)

print("Successfully processed:", len(processed))
print("Failed:", failed)
print("Output:", processed_dir)
```

```python
import os
import csv
import numpy as np
import librosa
import soundfile as sf

root = "/content/drive/MyDrive/FastSpeech2_Internship"

for language in ["English", "Telugu"]:
    base = os.path.join(root, language, "processed")
    audio_dir = os.path.join(base, "audio")
    mel_dir = os.path.join(base, "mel")
    energy_dir = os.path.join(base, "energy")

    os.makedirs(mel_dir, exist_ok=True)
    os.makedirs(energy_dir, exist_ok=True)

    with open(os.path.join(base, "metadata.csv"), encoding="utf-8") as f:
        rows = list(csv.DictReader(f))

    completed = 0

    for row in rows:
        filename = row["filename"]
        wav_path = os.path.join(audio_dir, filename)

        try:
            audio, sr = sf.read(wav_path, dtype="float32")

            if sr != 22050:
                raise ValueError(f"Unexpected sample rate: {sr}")

            if audio.ndim == 2:
                audio = audio.mean(axis=1)

            # Mel spectrogram
            mel = librosa.feature.melspectrogram(
                y=audio,
                sr=22050,
                n_fft=1024,
                hop_length=256,
                win_length=1024,
                n_mels=80,
                fmin=0,
                fmax=8000,
                center=False,
                power=1.0
            )

            log_mel = np.log(np.maximum(mel, 1e-5)).T

            # Frame-level energy
            frames = librosa.util.frame(
                np.pad(audio, (0, max(0, 1024 - len(audio)))),
                frame_length=1024,
                hop_length=256
            )

            energy = np.sqrt(np.sum(frames ** 2, axis=0))

            # Match energy length to mel frames
            energy = energy[:len(log_mel)]

            stem = os.path.splitext(filename)[0]

            np.save(os.path.join(mel_dir, stem + ".npy"),
                    log_mel.astype(np.float32))

            np.save(os.path.join(energy_dir, stem + ".npy"),
                    energy.astype(np.float32))

            completed += 1

        except Exception as e:
            print(language, filename, "Error:", str(e)[:150])

    print(language, "processed:", completed, "/", len(rows))
```

```python
import os
import csv
import numpy as np
import librosa
import soundfile as sf

root = "/content/drive/MyDrive/FastSpeech2_Internship"

for language in ["English", "Telugu"]:
    base = os.path.join(root, language, "processed")
    pitch_dir = os.path.join(base, "pitch")
    os.makedirs(pitch_dir, exist_ok=True)

    with open(os.path.join(base, "metadata.csv"), encoding="utf-8") as f:
        rows = list(csv.DictReader(f))

    completed = 0
    failed = 0

    for row in rows:
        filename = row["filename"]
        stem = os.path.splitext(filename)[0]

        try:
            audio, sr = sf.read(
                os.path.join(base, "audio", filename),
                dtype="float32"
            )

            if audio.ndim == 2:
                audio = audio.mean(axis=1)

            if sr != 22050:
                raise ValueError(f"Unexpected sample rate: {sr}")

            # Extract frame-level fundamental frequency
            pitch = librosa.yin(
                audio,
                fmin=65,
                fmax=500,
                sr=sr,
                frame_length=1024,
                hop_length=256,
                center=False
            )

            mel = np.load(os.path.join(base, "mel", stem + ".npy"))
            target_length = len(mel)

            # pitch length to mel length
            pitch = pitch[:target_length]

            if len(pitch) < target_length:
                pitch = np.pad(
                    pitch,
                    (0, target_length - len(pitch)),
                    mode="edge"
                )

            np.save(
                os.path.join(pitch_dir, stem + ".npy"),
                pitch.astype(np.float32)
            )

            completed += 1

        except Exception as e:
            failed += 1
            if failed <= 3:
                print(language, filename, "Error:", str(e)[:150])

    print(language, "pitch extracted:", completed, "failed:", failed)
```

```python
import os, glob
import numpy as np

root = "/content/drive/MyDrive/FastSpeech2_Internship"
p = f"{root}/English/processed"

fixed = 0
errors = []

for duration_path in sorted(glob.glob(f"{p}/duration/*.npy")):
    name = os.path.splitext(os.path.basename(duration_path))[0]

    try:
        duration = np.load(duration_path)
        target = int(duration.sum())

        if target <= 0:
            errors.append((name, "zero duration"))
            continue

        for feature in ["mel", "pitch", "energy"]:
            path = f"{p}/{feature}/{name}.npy"
            data = np.load(path)

            # Ensure all acoustic features match the aligned duration.
            if len(data) >= target:
                data = data[:target]
            else:
                pad_shape = [(0, target - len(data))]
                pad_shape += [(0, 0)] * (data.ndim - 1)
                data = np.pad(data, pad_shape, mode="constant")

            np.save(path, data)

        fixed += 1

    except Exception as e:
        errors.append((name, str(e)))

print("Processed recordings:", fixed)
print("Errors:", errors[:5])

matched = 0

for path in glob.glob(f"{p}/duration/*.npy"):
    name = os.path.splitext(os.path.basename(path))[0]
    target = int(np.load(path).sum())

    if all(
        len(np.load(f"{p}/{feature}/{name}.npy")) == target
        for feature in ["mel", "pitch", "energy"]
    ):
        matched += 1

print("Fully matched recordings:", matched)
```

```python
import os, glob, json
import numpy as np

root = "/content/drive/MyDrive/FastSpeech2_Internship"
p = f"{root}/English/processed"

os.makedirs(f"{p}/phoneme_pitch", exist_ok=True)
os.makedirs(f"{p}/phoneme_energy", exist_ok=True)

vocabulary = {"<PAD>": 0, "<UNK>": 1}
completed = 0
errors = []

for phone_path in sorted(glob.glob(f"{p}/phonemes/*.txt")):
    name = os.path.splitext(os.path.basename(phone_path))[0]

    try:
        with open(phone_path, encoding="utf-8") as f:
            phones = f.read().split()

        durations = np.load(f"{p}/duration/{name}.npy")
        pitch = np.load(f"{p}/pitch/{name}.npy")
        energy = np.load(f"{p}/energy/{name}.npy")
        mel = np.load(f"{p}/mel/{name}.npy")

        if len(phones) != len(durations):
            raise ValueError("Phoneme and duration counts differ")

        if not (
            int(durations.sum()) == len(mel)
            == len(pitch) == len(energy)
        ):
            raise ValueError("Acoustic feature lengths differ")

        for phone in phones:
            if phone not in vocabulary:
                vocabulary[phone] = len(vocabulary)

        phone_pitch = []
        phone_energy = []
        position = 0

        for d in durations:
            d = int(d)

            if d > 0:
                p_segment = pitch[position:position + d]
                e_segment = energy[position:position + d]

                voiced = p_segment[
                    np.isfinite(p_segment) & (p_segment > 0)
                ]

                phone_pitch.append(
                    float(voiced.mean()) if len(voiced) else 0.0
                )
                phone_energy.append(
                    float(e_segment.mean()) if len(e_segment) else 0.0
                )
            else:
                phone_pitch.append(0.0)
                phone_energy.append(0.0)

            position += d

        np.save(
            f"{p}/phoneme_pitch/{name}.npy",
            np.array(phone_pitch, dtype=np.float32)
        )
        np.save(
            f"{p}/phoneme_energy/{name}.npy",
            np.array(phone_energy, dtype=np.float32)
        )

        completed += 1

    except Exception as e:
        errors.append((name, str(e)))

with open(f"{p}/vocab.json", "w", encoding="utf-8") as f:
    json.dump(vocabulary, f, indent=2)

print("Processed recordings:", completed)
print("Vocabulary size:", len(vocabulary))
print("Errors:", errors[:5])
```

```python

import os
import json
import torch
import torch.nn as nn
import torch.nn.functional as F

root = "/content/drive/MyDrive/FastSpeech2_Internship"
processed = f"{root}/English/processed"

with open(f"{processed}/vocab.json", encoding="utf-8") as f:
    vocab = json.load(f)

class FastSpeech2Prototype(nn.Module):
    def __init__(self, vocab_size, hidden=192, n_mels=80):
        super().__init__()

        self.embedding = nn.Embedding(vocab_size, hidden, padding_idx=0)

        self.encoder = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(
                d_model=hidden,
                nhead=4,
                dim_feedforward=hidden * 4,
                dropout=0.1,
                batch_first=True
            ),
            num_layers=2,
            enable_nested_tensor=False
        )

        self.duration_predictor = nn.Linear(hidden, 1)
        self.pitch_predictor = nn.Linear(hidden, 1)
        self.energy_predictor = nn.Linear(hidden, 1)

        self.pitch_embedding = nn.Linear(1, hidden)
        self.energy_embedding = nn.Linear(1, hidden)

        self.decoder = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(
                d_model=hidden,
                nhead=4,
                dim_feedforward=hidden * 4,
                dropout=0.1,
                batch_first=True
            ),
            num_layers=2,
            enable_nested_tensor=False
        )

        self.mel_projection = nn.Linear(hidden, n_mels)

    def forward(self, phonemes, durations, pitch=None, energy=None):
        x = self.embedding(phonemes)
        x = self.encoder(x)

        log_duration_pred = self.duration_predictor(x).squeeze(-1)
        pitch_pred = self.pitch_predictor(x).squeeze(-1)
        energy_pred = self.energy_predictor(x).squeeze(-1)

        if pitch is None:
            pitch = pitch_pred
        if energy is None:
            energy = energy_pred

        x = (
            x
            + self.pitch_embedding(pitch.unsqueeze(-1))
            + self.energy_embedding(energy.unsqueeze(-1))
        )

        # Expand each phoneme according to its duration.
        expanded = torch.repeat_interleave(
            x.squeeze(0),
            durations.squeeze(0).long().clamp(min=0),
            dim=0
        )

        if expanded.shape[0] == 0:
            raise ValueError("All phoneme durations are zero")

        expanded = expanded.unsqueeze(0)

        decoded = self.decoder(expanded)
        mel = self.mel_projection(decoded)

        return mel, log_duration_pred, pitch_pred, energy_pred


device = "cuda" if torch.cuda.is_available() else "cpu"

model = FastSpeech2Prototype(
    vocab_size=len(vocab)
).to(device)

print("Device:", device)
print("Vocabulary size:", len(vocab))
print("Model parameters:", sum(p.numel() for p in model.parameters()))
print("Model created successfully!")
```

```python

import os, json, random
import numpy as np
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt

root = "/content/drive/MyDrive/FastSpeech2_Internship"
p = f"{root}/English/processed"
checkpoint_dir = f"{root}/English/checkpoints"
os.makedirs(checkpoint_dir, exist_ok=True)

with open(f"{p}/vocab.json", encoding="utf-8") as f:
    vocab = json.load(f)

with open(f"{p}/train_files.txt") as f:
    train_names = [x.strip() for x in f if x.strip()]

with open(f"{p}/val_files.txt") as f:
    val_names = [x.strip() for x in f if x.strip()]

def load_sample(name):
    with open(f"{p}/phonemes/{name}.txt", encoding="utf-8") as f:
        phones = f.read().split()

    ids = torch.tensor(
        [vocab.get(ph, vocab["<UNK>"]) for ph in phones],
        dtype=torch.long, device=device
    ).unsqueeze(0)

    durations = torch.tensor(
        np.load(f"{p}/duration/{name}.npy").astype(np.int64),
        device=device
    ).unsqueeze(0)

    pitch = torch.tensor(
        np.load(f"{p}/phoneme_pitch/{name}.npy"),
        dtype=torch.float32, device=device
    ).unsqueeze(0)

    energy = torch.tensor(
        np.load(f"{p}/phoneme_energy/{name}.npy"),
        dtype=torch.float32, device=device
    ).unsqueeze(0)

    mel = torch.tensor(
        np.load(f"{p}/mel/{name}.npy"),
        dtype=torch.float32, device=device
    ).unsqueeze(0)

    return ids, durations, pitch, energy, mel

def calculate_loss(name):
    ids, durations, pitch, energy, mel_target = load_sample(name)

    mel_pred, duration_pred, pitch_pred, energy_pred = model(
        ids, durations, pitch, energy
    )

    if mel_pred.shape != mel_target.shape:
        raise ValueError(
            f"{name}: predicted mel {mel_pred.shape}, "
            f"target mel {mel_target.shape}"
        )

    mel_loss = F.l1_loss(mel_pred, mel_target)

    duration_loss = F.mse_loss(
        duration_pred,
        torch.log1p(durations.float())
    )

    # Normalize targets for more stable prototype training.
    pitch_target = torch.log1p(pitch.clamp(min=0))
    energy_target = torch.log1p(energy.clamp(min=0))

    pitch_loss = F.mse_loss(pitch_pred, pitch_target)
    energy_loss = F.mse_loss(energy_pred, energy_target)

    total_loss = (
        mel_loss
        + 0.1 * duration_loss
        + 0.1 * pitch_loss
        + 0.1 * energy_loss
    )

    return total_loss

optimizer = torch.optim.Adam(model.parameters(), lr=0.0001)

epochs = 3
train_history = []
val_history = []

for epoch in range(epochs):
    model.train()
    random.shuffle(train_names)
    training_losses = []

    for i, name in enumerate(train_names):
        optimizer.zero_grad()

        loss = calculate_loss(name)

        if not torch.isfinite(loss):
            print("Skipping non-finite loss:", name)
            continue

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        training_losses.append(loss.item())

        if (i + 1) % 20 == 0:
            print(
                f"Epoch {epoch+1}/{epochs} | "
                f"Sample {i+1}/{len(train_names)} | "
                f"Loss {loss.item():.4f}"
            )

    model.eval()
    validation_losses = []

    with torch.no_grad():
        for name in val_names:
            loss = calculate_loss(name)
            if torch.isfinite(loss):
                validation_losses.append(loss.item())

    train_avg = float(np.mean(training_losses))
    val_avg = float(np.mean(validation_losses))

    train_history.append(train_avg)
    val_history.append(val_avg)

    print(
        f"Epoch {epoch+1} complete | "
        f"Train: {train_avg:.4f} | "
        f"Validation: {val_avg:.4f}"
    )

    torch.save({
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "epoch": epoch + 1,
        "vocab": vocab,
        "train_loss": train_avg,
        "val_loss": val_avg
    }, f"{checkpoint_dir}/english_latest.pth")

results_dir = f"{root}/Results/English"
os.makedirs(results_dir, exist_ok=True)

plt.figure(figsize=(7, 4))
plt.plot(range(1, epochs + 1), train_history, label="Training")
plt.plot(range(1, epochs + 1), val_history, label="Validation")
plt.xlabel("Epoch")
plt.ylabel("Total loss")
plt.title("English FastSpeech2 Prototype Training")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(f"{results_dir}/loss_curve.png", dpi=150)
plt.show()

print("Training finished!")
print("Checkpoint:", f"{checkpoint_dir}/english_latest.pth")
```

```python

import os
import json
import numpy as np
import pandas as pd

root = "/content/drive/MyDrive/FastSpeech2_Internship"
p = f"{root}/Telugu/processed"

metadata = pd.read_csv(f"{p}/metadata.csv")

vocab_te = {"<PAD>": 0, "<UNK>": 1}
fixed = 0
errors = []

for _, row in metadata.iterrows():
    name = os.path.splitext(str(row["filename"]))[0]

    try:
        # Use exactly the same splitting method as training.
        phonemes = str(row["phonemes"]).replace("_", " ").split()

        mel = np.load(f"{p}/mel/{name}.npy")
        pitch = np.load(f"{p}/pitch/{name}.npy")
        energy = np.load(f"{p}/energy/{name}.npy")

        count = len(phonemes)
        frames = len(mel)

        if count == 0 or frames < count:
            raise ValueError(
                f"Invalid count: {count} phonemes, {frames} frames"
            )

        if len(pitch) != frames or len(energy) != frames:
            raise ValueError("Mel, pitch, energy frame mismatch")

        # Approximate uniform phoneme durations.
        boundaries = np.linspace(
            0, frames, count + 1
        ).astype(int)

        durations = np.diff(boundaries).astype(np.int64)

        phoneme_pitch = np.array([
            np.mean(pitch[boundaries[i]:boundaries[i+1]])
            for i in range(count)
        ], dtype=np.float32)

        phoneme_energy = np.array([
            np.mean(energy[boundaries[i]:boundaries[i+1]])
            for i in range(count)
        ], dtype=np.float32)

        if not np.isfinite(phoneme_pitch).all() or not np.isfinite(phoneme_energy).all():
            raise ValueError("Non-finite pitch or energy")

        np.save(f"{p}/duration/{name}.npy", durations)
        np.save(f"{p}/phoneme_pitch/{name}.npy", phoneme_pitch)
        np.save(f"{p}/phoneme_energy/{name}.npy", phoneme_energy)

        with open(f"{p}/phonemes/{name}.txt", "w", encoding="utf-8") as f:
            f.write(" ".join(phonemes))

        for ph in phonemes:
            if ph not in vocab_te:
                vocab_te[ph] = len(vocab_te)

        fixed += 1

    except Exception as e:
        errors.append((name, str(e)))

with open(f"{p}/vocab.json", "w", encoding="utf-8") as f:
    json.dump(vocab_te, f, ensure_ascii=False, indent=2)

print("Recordings repaired:", fixed)
print("Vocabulary size:", len(vocab_te))
print("Errors:", errors[:5])

# Verify every recording before training.
mismatches = []

for name in train_te + val_te:
    with open(f"{p}/phonemes/{name}.txt", encoding="utf-8") as f:
        count = len(f.read().split())

    duration = np.load(f"{p}/duration/{name}.npy")
    mel = np.load(f"{p}/mel/{name}.npy")

    if len(duration) != count or duration.sum() != len(mel):
        mismatches.append(name)

print("Remaining mismatches:", len(mismatches))
```

```python

import os
import textwrap
import importlib.util
import torch

root = "/content/drive/MyDrive/FastSpeech2_Internship"
code_dir = f"{root}/Code"
os.makedirs(code_dir, exist_ok=True)

model_code = """
import torch
import torch.nn as nn

class FastSpeech2Prototype(nn.Module):
    def __init__(self, vocab_size, hidden=192, n_mels=80):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size, hidden, padding_idx=0
        )

        self.encoder = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(
                d_model=hidden,
                nhead=4,
                dim_feedforward=hidden * 4,
                dropout=0.1,
                batch_first=True
            ),
            num_layers=2,
            enable_nested_tensor=False
        )

        self.duration_predictor = nn.Linear(hidden, 1)
        self.pitch_predictor = nn.Linear(hidden, 1)
        self.energy_predictor = nn.Linear(hidden, 1)

        self.pitch_embedding = nn.Linear(1, hidden)
        self.energy_embedding = nn.Linear(1, hidden)

        self.decoder = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(
                d_model=hidden,
                nhead=4,
                dim_feedforward=hidden * 4,
                dropout=0.1,
                batch_first=True
            ),
            num_layers=2,
            enable_nested_tensor=False
        )

        self.mel_projection = nn.Linear(hidden, n_mels)

    def forward(self, phonemes, durations, pitch=None, energy=None):
        x = self.encoder(self.embedding(phonemes))

        log_duration_pred = self.duration_predictor(x).squeeze(-1)
        pitch_pred = self.pitch_predictor(x).squeeze(-1)
        energy_pred = self.energy_predictor(x).squeeze(-1)

        if pitch is None:
            pitch = pitch_pred

        if energy is None:
            energy = energy_pred

        x = (
            x
            + self.pitch_embedding(pitch.unsqueeze(-1))
            + self.energy_embedding(energy.unsqueeze(-1))
        )

        expanded = torch.repeat_interleave(
            x.squeeze(0),
            durations.squeeze(0).long().clamp(min=0),
            dim=0
        )

        if expanded.shape[0] == 0:
            raise ValueError("All phoneme durations are zero")

        mel = self.mel_projection(
            self.decoder(expanded.unsqueeze(0))
        )

        return (
            mel,
            log_duration_pred,
            pitch_pred,
            energy_pred
        )
"""

model_path = f"{code_dir}/model.py"

with open(model_path, "w", encoding="utf-8") as f:
    f.write(textwrap.dedent(model_code).strip() + "\n")

# Check syntax
with open(model_path, encoding="utf-8") as f:
    compile(f.read(), model_path, "exec")

# Verify saved architecture matches both trained checkpoints
spec = importlib.util.spec_from_file_location(
    "saved_fastspeech2_model", model_path
)
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)

for language in ["English", "Telugu"]:
    checkpoint_path = (
        f"{root}/{language}/checkpoints/"
        f"{language.lower()}_latest.pth"
    )

    checkpoint = torch.load(
        checkpoint_path,
        map_location="cpu",
        weights_only=False
    )

    weights = checkpoint["model_state_dict"]
    vocab_size = weights["embedding.weight"].shape[0]

    test_model = module.FastSpeech2Prototype(
        vocab_size=vocab_size
    )
    test_model.load_state_dict(weights, strict=True)

    print(language, "checkpoint compatibility: PASSED")

print("Model file saved:", model_path)
print("Python syntax check: PASSED")
```

```python

import os
import textwrap

root = "/content/drive/MyDrive/FastSpeech2_Internship"
code_dir = f"{root}/Code"
os.makedirs(code_dir, exist_ok=True)

train_code = r'''
import argparse
import json
import os
import random

import numpy as np
import torch
import torch.nn.functional as F

from model import FastSpeech2Prototype


def main(args):
    random.seed(42)
    torch.manual_seed(42)

    processed = os.path.join(args.root, args.language, "processed")
    checkpoint_dir = os.path.join(
        args.root, args.language, "checkpoints"
    )
    os.makedirs(checkpoint_dir, exist_ok=True)

    with open(os.path.join(processed, "vocab.json"), encoding="utf-8") as f:
        vocab = json.load(f)

    def read_split(filename):
        with open(os.path.join(processed, filename)) as f:
            return [line.strip() for line in f if line.strip()]

    train_names = read_split("train_files.txt")
    val_names = read_split("val_files.txt")

    device = "cuda" if torch.cuda.is_available() else "cpu"
    model = FastSpeech2Prototype(len(vocab)).to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=args.lr)

    def sample_loss(name):
        with open(
            os.path.join(processed, "phonemes", name + ".txt"),
            encoding="utf-8"
        ) as f:
            phones = f.read().split()

        ids = torch.tensor(
            [[vocab.get(ph, vocab["<UNK>"]) for ph in phones]],
            dtype=torch.long, device=device
        )

        def feature(folder, dtype):
            arr = np.load(
                os.path.join(processed, folder, name + ".npy")
            )
            return torch.tensor(
                arr, dtype=dtype, device=device
            ).unsqueeze(0)

        durations = feature("duration", torch.long)
        pitch = feature("phoneme_pitch", torch.float32)
        energy = feature("phoneme_energy", torch.float32)
        target_mel = feature("mel", torch.float32)

        if ids.shape[1] != durations.shape[1]:
            raise ValueError(f"{name}: phoneme/duration mismatch")

        mel, duration_pred, pitch_pred, energy_pred = model(
            ids, durations, pitch, energy
        )

        if mel.shape != target_mel.shape:
            raise ValueError(f"{name}: mel shape mismatch")

        return (
            F.l1_loss(mel, target_mel)
            + 0.1 * F.mse_loss(
                duration_pred, torch.log1p(durations.float())
            )
            + 0.1 * F.mse_loss(
                pitch_pred, torch.log1p(pitch.clamp(min=0))
            )
            + 0.1 * F.mse_loss(
                energy_pred, torch.log1p(energy.clamp(min=0))
            )
        )

    for epoch in range(args.epochs):
        model.train()
        random.shuffle(train_names)
        train_losses = []

        for name in train_names:
            optimizer.zero_grad()
            loss = sample_loss(name)

            if not torch.isfinite(loss):
                continue

            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            train_losses.append(loss.item())

        model.eval()
        val_losses = []

        with torch.no_grad():
            for name in val_names:
                loss = sample_loss(name)
                if torch.isfinite(loss):
                    val_losses.append(loss.item())

        train_avg = float(np.mean(train_losses))
        val_avg = float(np.mean(val_losses))

        print(
            f"Epoch {epoch+1}/{args.epochs}: "
            f"train={train_avg:.4f}, val={val_avg:.4f}"
        )

        torch.save({
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "epoch": epoch + 1,
            "vocab": vocab,
            "train_loss": train_avg,
            "val_loss": val_avg
        }, os.path.join(
            checkpoint_dir, args.language.lower() + "_latest.pth"
        ))


if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--language", choices=["English", "Telugu"],
                        required=True)
    parser.add_argument("--root", required=True)
    parser.add_argument("--epochs", type=int, default=3)
    parser.add_argument("--lr", type=float, default=0.0001)
    main(parser.parse_args())
'''

path = f"{code_dir}/train.py"

with open(path, "w", encoding="utf-8") as f:
    f.write(textwrap.dedent(train_code).strip() + "\n")

with open(path, encoding="utf-8") as f:
    compile(f.read(), path, "exec")

print("Training script saved:", path)
print("Python syntax check: PASSED")
```

```python

import os
import textwrap

root = "/content/drive/MyDrive/FastSpeech2_Internship"
code_dir = f"{root}/Code"
os.makedirs(code_dir, exist_ok=True)

preprocess_code = r'''
import argparse
import os
import glob

import librosa
import numpy as np
import soundfile as sf


SAMPLE_RATE = 22050
N_FFT = 1024
HOP_LENGTH = 256
WIN_LENGTH = 1024
N_MELS = 80


def preprocess_audio(wav_path, output_dir, name):
    audio, sr = librosa.load(
        wav_path, sr=SAMPLE_RATE, mono=True
    )

    # Log-mel magnitude spectrogram
    mel = librosa.feature.melspectrogram(
        y=audio,
        sr=SAMPLE_RATE,
        n_fft=N_FFT,
        hop_length=HOP_LENGTH,
        win_length=WIN_LENGTH,
        n_mels=N_MELS,
        fmin=0,
        fmax=8000,
        center=False,
        power=1.0
    )

    mel = np.log(np.maximum(mel, 1e-5)).T.astype(np.float32)

    # Frame-level pitch estimation
    pitch = librosa.yin(
        audio,
        fmin=65,
        fmax=800,
        sr=SAMPLE_RATE,
        frame_length=N_FFT,
        hop_length=HOP_LENGTH,
        center=False
    ).astype(np.float32)

    # Frame-level RMS energy
    energy = librosa.feature.rms(
        y=audio,
        frame_length=N_FFT,
        hop_length=HOP_LENGTH,
        center=False
    )[0].astype(np.float32)

    # Match feature lengths
    frames = min(len(mel), len(pitch), len(energy))
    mel = mel[:frames]
    pitch = pitch[:frames]
    energy = energy[:frames]

    for folder in ["mel", "pitch", "energy"]:
        os.makedirs(os.path.join(output_dir, folder), exist_ok=True)

    np.save(os.path.join(output_dir, "mel", name + ".npy"), mel)
    np.save(os.path.join(output_dir, "pitch", name + ".npy"), pitch)
    np.save(os.path.join(output_dir, "energy", name + ".npy"), energy)


def main(args):
    raw_dir = os.path.join(args.root, args.language, "raw_data")
    output_dir = os.path.join(
        args.root, args.language, "processed"
    )

    wav_files = sorted(glob.glob(os.path.join(raw_dir, "*.wav")))

    print("Language:", args.language)
    print("Audio files found:", len(wav_files))

    for i, wav_path in enumerate(wav_files):
        name = os.path.splitext(os.path.basename(wav_path))[0]
        preprocess_audio(wav_path, output_dir, name)

        if (i + 1) % 20 == 0:
            print("Processed:", i + 1)

    print("Acoustic feature extraction completed.")
    print("Note: phoneme extraction and alignment are separate steps.")


if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument(
        "--language",
        choices=["English", "Telugu"],
        required=True
    )
    parser.add_argument("--root", required=True)
    main(parser.parse_args())
'''

path = f"{code_dir}/preprocess.py"

with open(path, "w", encoding="utf-8") as f:
    f.write(textwrap.dedent(preprocess_code).strip() + "\n")

with open(path, encoding="utf-8") as f:
    compile(f.read(), path, "exec")

print("Preprocessing script saved:", path)
print("Python syntax check: PASSED")

print("\nCode files:")
for name in ["model.py", "train.py", "inference.py", "preprocess.py"]:
    file_path = f"{code_dir}/{name}"
    print(name, "FOUND" if os.path.exists(file_path) else "MISSING")
```

```python

import os
import csv
import subprocess
import soundfile as sf
from scipy.signal import resample_poly
from math import gcd

base = "/content/drive/MyDrive/FastSpeech2_Internship/Telugu"
raw = os.path.join(base, "raw_data")
processed = os.path.join(base, "processed")
audio_out = os.path.join(processed, "audio")

os.makedirs(audio_out, exist_ok=True)

with open(os.path.join(raw, "metadata.csv"), encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

success = 0
failed = []

with open(
    os.path.join(processed, "metadata.csv"),
    "w",
    newline="",
    encoding="utf-8"
) as f:

    writer = csv.writer(f)
    writer.writerow(["filename", "text", "phonemes"])

    for row in rows:
        try:
            filename = row["filename"]
            text = row["text"]

            audio, sr = sf.read(os.path.join(raw, filename))

            if audio.ndim > 1:
                audio = audio.mean(axis=1)

            if sr != 22050:
                g = gcd(sr, 22050)
                audio = resample_poly(
                    audio, 22050 // g, sr // g
                )

            new_name = os.path.splitext(filename)[0] + ".wav"

            sf.write(
                os.path.join(audio_out, new_name),
                audio,
                22050
            )

            result = subprocess.run(
                [
                    "espeak-ng", "-q", "-v", "te",
                    "-x", "--sep=_", text
                ],
                capture_output=True,
                text=True,
                check=True
            )

            phonemes = result.stdout.strip()

            if not phonemes:
                raise ValueError("Empty phoneme output")

            writer.writerow([new_name, text, phonemes])
            success += 1

        except Exception as e:
            failed.append((row["filename"], str(e)))

print("Successfully processed:", success)
print("Failed:", len(failed))
print("Output folder:", processed)

if failed:
    print("First errors:", failed[:5])
```

```python

import os
import csv
import numpy as np
import librosa

ROOT = "/content/drive/MyDrive/FastSpeech2_Internship"
SR = 22050
HOP = 256

for language in ["English", "Telugu"]:
    base = os.path.join(ROOT, language, "processed")
    audio_dir = os.path.join(base, "audio")
    mel_dir = os.path.join(base, "mels")
    feature_dir = os.path.join(base, "features")

    os.makedirs(feature_dir, exist_ok=True)

    with open(
        os.path.join(base, "metadata.csv"),
        encoding="utf-8"
    ) as f:
        rows = list(csv.DictReader(f))

    success = 0
    failed = []

    for row in rows:
        try:
            stem = os.path.splitext(row["filename"])[0]

            phonemes = split_phonemes(row["phonemes"])
            n_phonemes = len(phonemes)

            mel = np.load(
                os.path.join(mel_dir, stem + ".npy")
            )
            n_frames = mel.shape[1]

            if n_phonemes == 0 or n_frames == 0:
                raise ValueError("Empty phonemes or mel")

            # Uniform estimated phoneme durations
            durations = np.full(
                n_phonemes,
                n_frames // n_phonemes,
                dtype=np.int64
            )
            durations[:n_frames % n_phonemes] += 1

            y, _ = librosa.load(
                os.path.join(audio_dir, row["filename"]),
                sr=SR
            )

            # Frame-level pitch
            pitch = librosa.yin(
                y,
                fmin=65,
                fmax=600,
                sr=SR,
                frame_length=1024,
                hop_length=HOP
            )

            # Frame-level energy
            energy = librosa.feature.rms(
                y=y,
                frame_length=1024,
                hop_length=HOP
            )[0]

            # Match mel frame count
            pitch = librosa.util.fix_length(
                pitch, size=n_frames
            )
            energy = librosa.util.fix_length(
                energy, size=n_frames
            )

            # Convert frame-level values to phoneme-level
            pitch_ph = []
            energy_ph = []

            start = 0

            for duration in durations:
                end = start + int(duration)

                pitch_ph.append(
                    float(np.mean(pitch[start:end]))
                    if end > start else 0.0
                )
                energy_ph.append(
                    float(np.mean(energy[start:end]))
                    if end > start else 0.0
                )

                start = end

            np.savez(
                os.path.join(feature_dir, stem + ".npz"),
                durations=durations,
                pitch=np.asarray(pitch_ph, dtype=np.float32),
                energy=np.asarray(energy_ph, dtype=np.float32)
            )

            success += 1

        except Exception as e:
            failed.append((row["filename"], str(e)))

    print("\nLanguage:", language)
    print("Features created:", success)
    print("Failed:", len(failed))

    if failed:
        print("First errors:", failed[:3])
```

```python

import os
import csv
import json
import numpy as np
import torch
from torch.utils.data import Dataset, random_split

ROOT = "/content/drive/MyDrive/FastSpeech2_Internship"

class TTSDataset(Dataset):
    def __init__(self, language):
        self.base = os.path.join(ROOT, language, "processed")

        with open(
            os.path.join(self.base, "metadata.csv"),
            encoding="utf-8"
        ) as f:
            self.rows = list(csv.DictReader(f))

        with open(
            os.path.join(self.base, "vocab.json"),
            encoding="utf-8"
        ) as f:
            self.vocab = json.load(f)

    def __len__(self):
        return len(self.rows)

    def __getitem__(self, index):
        row = self.rows[index]
        stem = os.path.splitext(row["filename"])[0]

        phonemes = split_phonemes(row["phonemes"])

        ids = [
            self.vocab.get(p, 1)
            for p in phonemes
        ]

        mel = np.load(
            os.path.join(self.base, "mels", stem + ".npy")
        ).T.copy()

        with np.load(
            os.path.join(self.base, "features", stem + ".npz")
        ) as features:
            durations = features["durations"].copy()
            pitch = features["pitch"].copy()
            energy = features["energy"].copy()

        assert len(ids) == len(durations)
        assert len(ids) == len(pitch)
        assert len(ids) == len(energy)
        assert sum(durations) == len(mel)

        return {
            "phonemes": torch.tensor(ids, dtype=torch.long),
            "mel": torch.tensor(mel, dtype=torch.float32),
            "durations": torch.tensor(durations, dtype=torch.long),
            "pitch": torch.tensor(pitch, dtype=torch.float32),
            "energy": torch.tensor(energy, dtype=torch.float32)
        }

datasets_by_language = {}

for language in ["English", "Telugu"]:
    dataset = TTSDataset(language)

    train_set, val_set = random_split(
        dataset,
        [90, 10],
        generator=torch.Generator().manual_seed(42)
    )

    datasets_by_language[language] = {
        "dataset": dataset,
        "train": train_set,
        "val": val_set
    }

    sample = dataset[0]

    print("\nLanguage:", language)
    print("Training samples:", len(train_set))
    print("Validation samples:", len(val_set))
    print("Phonemes:", sample["phonemes"].shape)
    print("Durations:", sample["durations"].shape)
    print("Pitch:", sample["pitch"].shape)
    print("Energy:", sample["energy"].shape)
    print("Mel:", sample["mel"].shape)
    print("Duration sum matches mel:",
          sample["durations"].sum().item() == len(sample["mel"]))
```

```python

import torch
import torch.nn as nn

class FastSpeech2Prototype(nn.Module):
    def __init__(self, vocab_size, hidden=96, n_mels=80):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size, hidden, padding_idx=0
        )

        self.encoder = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(
                d_model=hidden,
                nhead=4,
                dim_feedforward=hidden * 2,
                dropout=0.1,
                batch_first=True
            ),
            num_layers=1,
            enable_nested_tensor=False
        )

        self.duration_predictor = nn.Linear(hidden, 1)
        self.pitch_predictor = nn.Linear(hidden, 1)
        self.energy_predictor = nn.Linear(hidden, 1)

        self.pitch_embedding = nn.Linear(1, hidden)
        self.energy_embedding = nn.Linear(1, hidden)

        self.decoder = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(
                d_model=hidden,
                nhead=4,
                dim_feedforward=hidden * 2,
                dropout=0.1,
                batch_first=True
            ),
            num_layers=1,
            enable_nested_tensor=False
        )

        self.mel_projection = nn.Linear(hidden, n_mels)

    def forward(
        self, phonemes, durations,
        pitch=None, energy=None
    ):
        x = self.embedding(phonemes)
        x = self.encoder(x)

        duration_pred = self.duration_predictor(x).squeeze(-1)
        pitch_pred = self.pitch_predictor(x).squeeze(-1)
        energy_pred = self.energy_predictor(x).squeeze(-1)

        if pitch is None:
            pitch = pitch_pred

        if energy is None:
            energy = energy_pred

        x = (
            x
            + self.pitch_embedding(pitch.unsqueeze(-1))
            + self.energy_embedding(energy.unsqueeze(-1))
        )

        expanded = torch.repeat_interleave(
            x.squeeze(0),
            durations.squeeze(0).long().clamp(min=0),
            dim=0
        )

        if expanded.shape[0] == 0:
            raise ValueError("All phoneme durations are zero")

        mel = self.mel_projection(
            self.decoder(expanded.unsqueeze(0))
        )

        return (
            mel,
            duration_pred,
            pitch_pred,
            energy_pred
        )

print("FastSpeech2Prototype model defined successfully")

for language in ["English", "Telugu"]:
    vocab_size = len(
        datasets_by_language[language]["dataset"].vocab
    )

    model = FastSpeech2Prototype(vocab_size)

    parameters = sum(
        p.numel() for p in model.parameters()
    )

    print(
        language,
        "| Vocabulary:", vocab_size,
        "| Parameters:", parameters
    )
```

```python

import os
import time
import json
import torch
import torch.nn.functional as F

torch.set_num_threads(2)

language = "English"
device = torch.device("cpu")

dataset_info = datasets_by_language[language]
train_set = dataset_info["train"]
val_set = dataset_info["val"]

vocab_size = len(dataset_info["dataset"].vocab)

model = FastSpeech2Prototype(vocab_size).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

checkpoint_dir = os.path.join(
    ROOT, language, "checkpoints"
)
os.makedirs(checkpoint_dir, exist_ok=True)

history = {
    "train_loss": [],
    "val_loss": []
}

def calculate_loss(model, sample):
    phonemes = sample["phonemes"].unsqueeze(0).to(device)
    durations = sample["durations"].unsqueeze(0).to(device)
    pitch = sample["pitch"].unsqueeze(0).to(device)
    energy = sample["energy"].unsqueeze(0).to(device)
    mel_target = sample["mel"].unsqueeze(0).to(device)

    mel_pred, duration_pred, pitch_pred, energy_pred = model(
        phonemes, durations, pitch, energy
    )

    mel_loss = F.l1_loss(mel_pred, mel_target)

    duration_loss = F.mse_loss(
        duration_pred,
        torch.log1p(durations.float())
    )

    pitch_loss = F.mse_loss(
        pitch_pred,
        torch.log1p(pitch.clamp(min=0))
    )

    energy_loss = F.mse_loss(
        energy_pred,
        torch.log1p(energy.clamp(min=0))
    )

    return (
        mel_loss
        + 0.1 * duration_loss
        + 0.1 * pitch_loss
        + 0.1 * energy_loss
    )

start_time = time.time()

for epoch in range(3):
    model.train()
    train_loss = 0.0

    for i in range(len(train_set)):
        sample = train_set[i]

        optimizer.zero_grad()

        loss = calculate_loss(model, sample)
        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(), 1.0
        )

        optimizer.step()
        train_loss += loss.item()

        if (i + 1) % 10 == 0:
            print(
                f"Epoch {epoch+1} | "
                f"Sample {i+1}/{len(train_set)} | "
                f"Loss {loss.item():.4f}",
                flush=True
            )

    train_loss /= len(train_set)

    model.eval()
    val_loss = 0.0

    with torch.no_grad():
        for i in range(len(val_set)):
            val_loss += calculate_loss(
                model, val_set[i]
            ).item()

    val_loss /= len(val_set)

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)

    checkpoint = {
        "epoch": epoch + 1,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "history": history,
        "vocab_size": vocab_size,
        "hidden": 96
    }

    checkpoint_path = os.path.join(
        checkpoint_dir, "english_latest.pth"
    )

    torch.save(checkpoint, checkpoint_path)

    with open(
        os.path.join(checkpoint_dir, "history.json"),
        "w"
    ) as f:
        json.dump(history, f, indent=2)

    print(
        f"\nEpoch {epoch+1}/3 completed | "
        f"Train Loss: {train_loss:.4f} | "
        f"Validation Loss: {val_loss:.4f}",
        flush=True
    )
    print("Checkpoint saved:", checkpoint_path, flush=True)

print(
    "\nTotal training time (minutes):",
    round((time.time() - start_time) / 60, 2)
)
```

```python

import os
import json
import numpy as np
import torch
import librosa
import soundfile as sf
import subprocess

ROOT = "/content/drive/MyDrive/FastSpeech2_Internship"
torch.set_num_threads(2)

test_sentences = {
    "English": {
        "E1": "The antidisestablishmentarianism movement gained considerable momentum during the parliamentary proceedings.",
        "E2a": "The orchestra's conduct during the ceremony was praised by the audience.",
        "E2b": "The researchers will conduct an experiment to evaluate the proposed method.",
        "E3": "The rhythm of the orchestra changed unexpectedly.",
        "E4": "The meeting will start at 10:45 AM on 12 November 2026, so andaroo should arrive on time."
    },
    "Telugu": {
        "T1": "న్యూ రోఫార్మ కోలాజికల్ పరిశోధనలో నూతన ఔషధాల ప్రభావాన్ని విశ్లేషించారు.",
        "T2a": "తాడుపై అతని పట్టు చాలా బలంగా ఉంది.",
        "T2b": "ఆమె పట్టు చీరను ధరించి వేడుకకు వచ్చింది.",
        "T3": "పులి అడవిలో నిశ్శబ్దంగా నడిచింది.",
        "T4": "ఈ సమావేశం 2026 నవంబర్ 12వ తేదీ ఉదయం 10:45 గంటలకు ప్రారంభమవుతుంది, సమయానికి సిద్ధంగా ఉంచండి."
    }
}

def text_to_phonemes(text, language):
    voice = "en-us" if language == "English" else "te"

    result = subprocess.run(
        ["espeak-ng", "-q", "-v", voice, "-x", "--sep=_", text],
        capture_output=True,
        text=True,
        check=True
    )

    return [
        p.strip()
        for p in result.stdout.replace("\n", " ").split("_")
        if p.strip()
    ]

for language, sentences in test_sentences.items():
    print("\nProcessing:", language, flush=True)

    base = os.path.join(ROOT, language)
    checkpoint_path = os.path.join(
        base, "checkpoints",
        "english_latest.pth" if language == "English"
        else "telugu_latest.pth"
    )

    with open(
        os.path.join(base, "processed", "vocab.json"),
        encoding="utf-8"
    ) as f:
        vocab = json.load(f)

    checkpoint = torch.load(
        checkpoint_path,
        map_location="cpu",
        weights_only=True
    )

    model = FastSpeech2Prototype(
        vocab_size=len(vocab),
        hidden=96
    )

    model.load_state_dict(checkpoint["model_state_dict"])
    model.eval()

    output_dir = os.path.join(
        ROOT, "Results", language, "test_generated_audios"
    )
    os.makedirs(output_dir, exist_ok=True)

    for test_id, text in sentences.items():
        try:
            phonemes = text_to_phonemes(text, language)
            ids = [vocab.get(p, 1) for p in phonemes]

            if not ids:
                raise ValueError("No phonemes generated")

            phoneme_tensor = torch.tensor(
                [ids], dtype=torch.long
            )

            with torch.no_grad():
                x = model.encoder(
                    model.embedding(phoneme_tensor)
                )

                log_duration = model.duration_predictor(
                    x
                ).squeeze(-1)

                pitch_pred = model.pitch_predictor(
                    x
                ).squeeze(-1)

                energy_pred = model.energy_predictor(
                    x
                ).squeeze(-1)

                durations = torch.expm1(
                    log_duration.clamp(0, 4)
                ).round().long().clamp(1, 20)

                pitch = torch.expm1(
                    pitch_pred.clamp(0, 8)
                ).clamp(0, 800)

                energy = torch.expm1(
                    energy_pred.clamp(0, 5)
                ).clamp(0, 100)

                mel_pred, _, _, _ = model(
                    phoneme_tensor,
                    durations,
                    pitch,
                    energy
                )

            mel = mel_pred.squeeze(0).numpy().T

            # Convert predicted log-mel to mel power
            mel_power = np.exp(
                np.clip(mel, -12, 8)
            )

            audio = librosa.feature.inverse.mel_to_audio(
                mel_power,
                sr=22050,
                n_fft=1024,
                hop_length=256,
                win_length=1024,
                n_iter=16
            )

            sf.write(
                os.path.join(output_dir, test_id + ".wav"),
                audio,
                22050
            )

            np.save(
                os.path.join(output_dir, test_id + "_mel.npy"),
                mel
            )
            np.save(
                os.path.join(output_dir, test_id + "_duration.npy"),
                durations.squeeze(0).numpy()
            )
            np.save(
                os.path.join(output_dir, test_id + "_pitch.npy"),
                pitch.squeeze(0).numpy()
            )
            np.save(
                os.path.join(output_dir, test_id + "_energy.npy"),
                energy.squeeze(0).numpy()
            )

            print(
                test_id,
                "saved | Duration:",
                round(len(audio) / 22050, 2),
                "seconds",
                flush=True
            )

        except Exception as e:
            print(test_id, "ERROR:", str(e), flush=True)

print("\nAudio generation finished")
```

```python
!pip install -q reportlab

import os
import json
import glob
from pathlib import Path

from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer,
    Table, TableStyle, Image, PageBreak
)
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib.enums import TA_CENTER
from reportlab.lib.units import inch

ROOT = Path("/content/drive/MyDrive/FastSpeech2_Internship")
report_dir = ROOT / "Report"
report_dir.mkdir(parents=True, exist_ok=True)

pdf_path = report_dir / "Technical_Report.pdf"

styles = getSampleStyleSheet()
story = []

def heading(text):
    story.append(Paragraph(text, styles["Heading2"]))
    story.append(Spacer(1, 8))

def paragraph(text):
    story.append(Paragraph(text, styles["BodyText"]))
    story.append(Spacer(1, 10))

story.append(
    Paragraph(
        "English and Telugu FastSpeech2-Style TTS Prototype",
        styles["Title"]
    )
)
story.append(Spacer(1, 18))

paragraph(
    "Technical Report | CPU-Based Proof-of-Concept Implementation"
)

heading("1. Project Objective")
paragraph(
    "The objective is to develop an English and Telugu "
    "text-to-speech prototype using a "
    "FastSpeech2-style architecture."
)

heading("2. Dataset")
data = [
    ["Language", "Source", "Train", "Validation"],
    ["English", "ai4bharat/Svarah", "90", "10"],
    ["Telugu", "SPRINGLab/IndicTTS_Telugu", "90", "10"]
]
table = Table(data, colWidths=[85, 190, 65, 75])
table.setStyle(TableStyle([
    ("BACKGROUND", (0,0), (-1,0), colors.lightgrey),
    ("GRID", (0,0), (-1,-1), 0.5, colors.grey),
    ("FONTSIZE", (0,0), (-1,-1), 8),
    ("VALIGN", (0,0), (-1,-1), "MIDDLE"),
]))
story.append(table)
story.append(Spacer(1, 14))

heading("3. Preprocessing")
paragraph(
    "Audio was resampled to 22050 Hz. "
    "80-bin log-mel spectrograms were extracted using "
    "1024-point FFT and 256-sample hop length. "
    "Phonemes were obtained using espeak-ng. "
    "Durations were estimated by approximately uniform "
    "allocation rather than forced alignment. "
    "Pitch and energy features were extracted and "
    "aggregated to phoneme level."
)

heading("4. Model Architecture")
paragraph(
    "This simplified model consists of phoneme embeddings, "
    "a Transformer encoder, duration/pitch/energy predictors, "
    "pitch and energy embeddings, a duration-based length "
    "regulator, a Transformer decoder and an 80-bin "
    "mel-spectrogram projection. "
    "The hidden dimension is 96, with one encoder layer "
    "and one decoder layer. "
)

heading("5. Training Configuration")
paragraph(
    "Training used PyTorch on CPU, Adam optimizer, "
    "learning rate 0.001, batch size 1, "
    "three epochs per language and a 90/10 split. "
    "The training objective combined mel reconstruction "
    "loss with duration, pitch and energy prediction losses."
)

heading("6. Training Results")

for language in ["English", "Telugu"]:
    history_path = (
        ROOT / language / "checkpoints" / "history.json"
    )
    history = json.loads(
        history_path.read_text(encoding="utf-8")
    )

    paragraph(f"<b>{language}</b>")

    rows = [["Epoch", "Train Loss", "Validation Loss"]]
    for i, (train, val) in enumerate(
        zip(history["train_loss"], history["val_loss"]), 1
    ):
        rows.append([
            str(i),
            f"{train:.4f}",
            f"{val:.4f}"
        ])

    t = Table(rows, colWidths=[100, 130, 130])
    t.setStyle(TableStyle([
        ("BACKGROUND", (0,0), (-1,0), colors.lightgrey),
        ("GRID", (0,0), (-1,-1), 0.5, colors.grey),
        ("FONTSIZE", (0,0), (-1,-1), 9),
    ]))
    story.append(t)
    story.append(Spacer(1, 12))

    for filename in [
        "training_loss.png", "validation_loss.png"
    ]:
        image_path = ROOT / "Results" / language / filename
        if image_path.exists():
            story.append(
                Image(str(image_path), width=5*inch, height=2.8*inch)
            )
            story.append(Spacer(1, 10))

heading("7. Inference and Generated Audio")
paragraph(
    "Five test utterances were synthesized for each language. "
    "Predicted mel-spectrograms were converted to WAV audio "
    "using Griffin-Lim reconstruction. "
    "Duration, pitch and energy prediction plots were saved "
    "for all ten test utterances. "
    "Audio generation was completed, but intelligibility "
    "and naturalness have not been formally evaluated."
)

heading("8. Limitations")
paragraph(
    "This implementation uses 100 samples per language, "
    "approximate phoneme alignment, a simplified model, "
    "and Griffin-Lim instead of a trained neural vocoder. "
    "It does not satisfy a multi-hour training-data "
    "requirement. No MOS, WER or CER evaluation was conducted. "
    "The generated speech may contain artifacts or "
    "unintelligible segments."
)

heading("9. Conclusion")
paragraph(
    "The project successfully builds an end-to-end bilingual "
    "FastSpeech2-style prototype with preprocessing, "
    "That covers model training, checkpointing, mel prediction, "
    "audio reconstruction and diagnostic plots. "
    "Further work should use larger aligned datasets, "
    "more complete FastSpeech2 training and a neural vocoder."
)

doc = SimpleDocTemplate(
    str(pdf_path),
    pagesize=(595, 842),
    rightMargin=45,
    leftMargin=45,
    topMargin=45,
    bottomMargin=45
)

doc.build(story)

print("Technical report saved:", pdf_path)

print("\nFINAL FILE VERIFICATION")

for language in ["English", "Telugu"]:
    result_dir = ROOT / "Results" / language
    audio_dir = result_dir / "test_generated_audios"

    wav_count = len(list(audio_dir.glob("*.wav")))
    plot_count = len(list(
        (result_dir / "variance_adaptor_plots").rglob("*.png")
    ))

    checkpoint_name = (
        "english_latest.pth"
        if language == "English"
        else "telugu_latest.pth"
    )

    checkpoint_exists = (
        ROOT / language / "checkpoints" / checkpoint_name
    ).exists()

    print(
        language,
        "| WAV:", wav_count,
        "| Variance plots:", plot_count,
        "| Checkpoint:", checkpoint_exists
    )

print("Report exists:", pdf_path.exists())
print("README exists:", (ROOT / "README.md").exists())
print("Requirements exists:", (ROOT / "requirements.txt").exists())
print("\ncompleted")
```

```python

from pathlib import Path
import zipfile

root = Path("/content/drive/MyDrive/FastSpeech2_Internship")
zip_path = root.parent / "FastSpeech2_Internship_Submission.zip"

with zipfile.ZipFile(
    zip_path, "w", compression=zipfile.ZIP_DEFLATED,
    compresslevel=3
) as z:
    for file in root.rglob("*"):
        if not file.is_file():
            continue

        if file.suffix == ".zip":
            continue

        relative_path = file.relative_to(root)
        z.write(file, arcname=str(relative_path))

print("ZIP created:", zip_path)
print("ZIP size (MB):", round(zip_path.stat().st_size / 1024**2, 2))

with zipfile.ZipFile(zip_path) as z:
    names = z.namelist()
    notebooks = [n for n in names if n.endswith(".ipynb")]
    print("Notebooks:", notebooks)
    print("Files in ZIP:", len(names))

    required = [
        "README.md",
        "requirements.txt",
        "Report/Technical_Report.pdf",
        "English/checkpoints/english_latest.pth",
        "Telugu/checkpoints/telugu_latest.pth"
    ]

    for name in required:
        print(name, "|", "OK" if name in names else "MISSING")
```
